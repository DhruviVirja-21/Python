### 15) WAP to check whether the given number is an Armstrong Number or not.

```python
n=input('enter n:')
r=1
sum=0
size=len(n)
n=int(n)
temp=n
while n!=0:
    r=n%10
    sum=sum+(r**size)
    n=n//10   
if(sum==temp):
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
    
    
```

### 16) WAP to print all the perfect numbers between 1 to n.

```python
n=int(input('enter n:'))
m=int(input('enter m:'))
i=n
sum=0
while i<m:
    j=1
    while j<i:
        if(i%j==0):
            sum=sum+j
        j+=1  
    if(i==sum):
        print(i)
    sum=0    
    i+=1
```
