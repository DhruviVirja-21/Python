### 01) WAP to find their union, intersection, and difference of the given two sets.

```

```

### 06) WAP to Sort a dictionary in ascending order of its values.
Hint: in sorted(), use udf for key parameter


**Sample Input:**
```
{'Math': 70, 'Physics': 85, 'Chemistry': 60}

```
**Sample Output:**
```
{'Chemistry': 60, 'Math': 70, 'Physics': 85}
```

````python
def get_value(item):
    return item[1]  

data = {'Math': 70, 'Physics': 85, 'Chemistry': 60}

sorted_dict=dict(sorted(data.items(), key=get_value))

print(sorted_dict)
````

### 07) WAP to find the keys having maximum and minimum values in a dictionary.

```
Min Value Key: C

```

````python
l1=list(input("Enter list keys: ").split())
l2=list(map(list, input("Enter list values: ").split()))
data=dict(zip(l1,l2))

maxkey = max(data, key=data.get)
minkey = min(data, key=data.get)

print("Max Value Key:", maxkey)
print("Min Value Key:", minkey)
````

### 11) WAP to find the student with the highest total marks, given student names and their marks.

**Sample Input:**
```
{'Amit': [70, 80, 90], 'Riya': [85, 75, 80], 'Neha': [60, 70, 65]}
```
**Sample Output:**
```
Topper: Amit
```

````python
l1 = list(input("Enter student names: ").split())
n = int(input("Enter number of marks per student: "))
d = {}

for name in l1:
    marks = list(map(int, input(f"Enter marks of {name}: ").split()))
    d[name] = marks
    
max_total=0
topper = ""

for name in d:
    total = sum(d[name])
    if total > max_total:
        max_total = total
        topper = name

print("Topper:", topper)

````
