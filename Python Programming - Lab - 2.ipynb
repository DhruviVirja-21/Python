### 03) WAP to accept three integers and display the numbers, their sum, and average.
#### Input: 10 20 30
#### Output: 
##### Numbers: 10 20 30
##### Sum: 60
##### Average: 20.0

```python
n1 = int(input("Enter 1 number:"))
n2 = int(input("Enter 2 number:"))
n3 = int(input("Enter 3 number:"))

sum = n1+n2+n3
avg = sum/3
print("Numbers: ",n1,n2,n3,sep=' ')
print("sum: ",sum)
print("avg: ",avg)
```

### 12) WAP to accept date in dd mm yyyy format and display in multiple formats.
