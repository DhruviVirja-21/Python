### 04) WAP to find the longest word(s) in a file.

```python
fp = open("abc.txt", "r")
l = 0
li = []
for i in fp.read().split():
    if len(i) > l:
        l = len(i)
        li = [i]      
    elif len(i) == l:
        li.append(i)

print(l)
print(li)
fp.close()
```

### 05) WAP to count the no. of lines, words and characters in a given text file.

```python
fp = open("abc.txt", "r")
data = fp.read()
chars = 0
for ch in data:
    if ch != " " and ch != "\n":
        chars += 1

print("lines =", data.count("\n") + 1)
print("words =", len(data.split()))
print("characters =", chars)

fp.close()
```

### 09) WAP to get the score of five subjects from the user, store them in a file. Fetch those marks and find the highest score.

```python
l=list(input("enter 5 Subjects Mark:").split("\n"))
max_mark="0"
with open("student.txt","r+")as fp:
    fp.writelines(l)
    fp.seek(0,0)
    for i in fp.read().split():
        if i > max_mark:
            max_mark=i
    print(f"max_mark={max_mark}")        
    fp.close()
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
fp = open("primenumbers.txt", "w")
count = 0
num = 2
while count < 100:
    for i in range(2, num):
        if num % i == 0:
            break
    else:
        fp.write(str(num) + "\n")
        count += 1
    num += 1

fp.close()
```
