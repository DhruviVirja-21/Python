### 14) WAP to extract elements with frequency greater than K.

```python
l = list(map(int, input().split()))
k = int(input("Enter K: "))

result = []

for i in l:
    if l.count(i) > k and i not in result:
        result.append(i)

print("Elements with frequency greater than K:", result)
```

### 16) WAP to create a new list (fruit whose name starts with 'b') from the list of fruits given by user.

```python
fruits = input("Enter fruits: ").split()
b_fruits = [f for f in fruits if f.lower().startswith('b')]
print("Fruits starting with 'b':", b_fruits)
```

### 17) WAP to create a list of common elements from given two lists.

```python
l1 = list(map(int,input("Enter values sep by space").split()))
l2 = list(map(int,input("Enter values sep by space").split()))
new=[]
for i in l1:
    for j in l2:
        if i==j:
            new.append(i)
            i=0
new
```
